Download the dataset


```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("miadul/kidney-function-health-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
df= pd.read_csv("kidney_dataset.csv")
df.head(10)
```

Visualization


```python
plt.scatter(df.Age,df.CKD_Status,color='red',marker='o')
plt.xlabel('Age')
plt.ylabel('CKD_Status')
plt.show()
```

Implementing Linear Regression Algorithm


```python
reg= linear_model.LinearRegression()
reg.fit(df[['Age']],df.CKD_Status)
```

Prediction

```python
reg.predict([[60]])

```

Coefficient

```python
reg.coef_
```

Intercept

```python
reg.intercept_
```
